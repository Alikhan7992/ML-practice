```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
car_data = pd.read_csv(r"ford.csv")
```

```python
car_data.head()
```

```python
car_data.shape
```

```python
car_data.isnull().sum()
```

```python
x = car_data.drop(columns=['price'])
y = car_data['price']
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
x.columns
```

```python
xoencode = x
```

```python
xoencode = pd.get_dummies(x,columns = ['model','transmission','fuelType'],drop_first = True)
```

```python
xoencode = xoencode.astype(int)
```

```python
xoencode
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
xlencode = x.copy()
```

```python
colm = ['model','transmission','fuelType']
for i in colm:
    l_encode = LabelEncoder()
    xlencode[i] = l_encode.fit_transform(xlencode[i])
```

```python
xlencode
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
xoencode[numerical_cols] = scaler.fit_transform(xoencode[numerical_cols])
```

```python
xoencode
```

```python
xlencode[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(xlencode[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
xlencode
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(xoencode,y,test_size=0.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_test,y_pred)*100
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r2)
```
