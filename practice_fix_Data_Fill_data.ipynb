# Now we are going to work for numeric data
**using simple imputer **

```python
bank_loan.select_dtypes(include= 'float64').columns
```

```python
mean_imputer = SimpleImputer(strategy='mean')
```

```python
fill_values = mean_imputer.fit_transform(bank_loan[['CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term',
       'Credit_History']])
```

```python
fill_values
```

```python
Fill_numeric_values = pd.DataFrame(fill_values,columns=bank_loan.select_dtypes(include='float64').columns)
```
