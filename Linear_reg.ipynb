```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
salary_data = pd.read_csv(r'SalaryData.csv')
```

```python
salary_data.head(3)
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x='YearsExperience', y='Salary', data = salary_data)
```

```python
x = salary_data[['YearsExperience']]
y = salary_data['Salary']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.20,random_state=40)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict([[1.1]])
```

```python
model.score(x_test, y_test)*100
```

```python
model.coef_
```

```python
model.intercept_
```

```python
9268.66191159 * 1.1 + 27531.140299387596
```

```python
pre = model.predict(x)
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x='YearsExperience', y='Salary', data = salary_data)
plt.plot(salary_data['YearsExperience'],pre, c = "red")
plt.legend(['Org Data' , 'Predict Data'])
plt.show()
```
